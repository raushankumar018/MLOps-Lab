```python
# Cell 1: Import Libraries
import pandas as pd
import numpy as np

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Cell 2: Upload Dataset
uploaded = files.upload()
```

```python
# Cell 3: Load Dataset
df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
# Cell 4: Dataset Information
print("Shape:", df.shape)

df.info()
```

```python
# Cell 5: Missing Values
print(df.isnull().sum())
```

```python
# a) Handle Missing Values
# Cell 6
# Fill Age with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill Embarked with most frequent value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin because most values are missing
df = df.drop("Cabin", axis=1)

print(df.isnull().sum())
```

```python
# b) Encode Categorical Variables
# Cell 7
encoder = LabelEncoder()

df["Sex"] = encoder.fit_transform(df["Sex"])

df["Embarked"] = encoder.fit_transform(df["Embarked"])
```

```python
# c) Normalize / Scale Data

# First remove text columns.

# Cell 8
df = df.drop(["Name","Ticket"], axis=1)

df.head()
```

```python
# Cell 9
X = df.drop("Survived", axis=1)

y = df["Survived"]
```

```python
# Cell 10
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

```python
# d) Split Dataset
# Cell 11
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
# e) Automate using Pipeline
# Cell 12
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

X_pipeline = pipeline.fit_transform(X)

print("Pipeline Applied Successfully!")
```

```python
# f) Save Processed Dataset
# Cell 13
processed_df = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

processed_df["Survived"] = y.values

processed_df.to_csv(
    "Titanic_Preprocessed.csv",
    index=False
)
```
